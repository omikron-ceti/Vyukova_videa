# Podmínky
Často se nám při psaní programu stane, že potřebujeme danou úlohu řešit různými způsoby, podle nějakého kritéria.  
Příklad: **řešení kvadratické rovnice:**  
 <img src="img/kvadraticka_rovnice_diagram.jpg" width="400 px"/>


K uskutečnění takovýchto rozhodování nám v Pythonu slouží **podmíněné příkazy**. Struktura podmíněného příkazu vypadá takto:
```python
if podmínka: # podmínka se vyhodnotí jako True nebo False
    něco_udělej()
    a_klidně_se_může_provést_víc_věcí()
else: 
    udělej_něco_jiného() # toto se provede, pokud podmínka není splněna
další_kód() # neodsazený kód už není součástí rozhodovacího řetězce a provede se vždy, bez ohledu na vyhodnocení podmínky
```
**Poznámka: `bool` je datový typ, který může nabývat hodnot `True` a `False`.** 

````python
a = 42
# a = -42

if a > 0: # 42 > 0, tedy podmínka ja vyhodnocena jako True
    print("číslo je kladné")
else:
    print("číslo není kladné") # tohle by se provedlo jen kdyby nebyla splněná podmínka

print("číslo se rovná", a) # tohle už není součástí rozhodovacího řetězce a provede se vždy
````

Podmínky jde vnořovat jednu do druhé, o správnou interpretaci kódu se postará odsazení

````python
a = -42
if a > 0:
    print("číslo je kladné")
    print("číslo se rovná", a)
else:
    if a == 0:
        print("číslo se rovná 0")
    else:
        print("číslo je záporné")
        print("číslo se rovná", a)
````

Aby se vnořování nevymyklo kontrole a nevznikl vám šílený [spaghetti code](https://www.geeksforgeeks.org/software-engineering/spaghetti-code/), může se hodit příkaz `elif` následovaný podmínkou. Příkazů `elif` můžete mít v jednom rozhodovacím řetězci kolik chcete. 

````python
a = 0
if a > 0:
    print("číslo je kladné")
    print("číslo se rovná", a)
elif a == 0:
    print("číslo se rovná 0")
else:
    print("číslo je záporné")
    print("číslo se rovná", a)
````

Pár důležitých upozornění:

## Za `else` se nikdy nikdy NIKDY nepíše podmínka!!!
Hodilo by vám to syntax error. Pro řetězení podmínek se používá `elif`!  
 
### Takhle NE!
```python
x = 42
if x < 420: 
  něco()
else x > 69:
  něco_jiného()
```

### Takhle TAKY NE:
(Python není C)
```python
x = 42
if x < 420: 
  něco()
else if x > 69:
  něco_jiného()
```

### Takhle ANO: 
```python
x = 42
if x < 420: 
  něco()
elif x > 69:
  něco_jiného()
```

---
## Příkazy `elif` a `else` nemůžou stát samostatně!!!
Rozhodovací řetězec musí vždy začínat příkazem `if`!

### Takhle NE!
```python
x = 42
else x > 16:
  něco()
```

### Takhle TAKY NE:
(Python není C)
```python
x = 42
elif x < 420: 
  něco()
else:
  něco_jiného()
```
---
## Použití `elif` a `else` je nepovinné
### Tohle je SPRÁVNĚ:
```python
x = 42
if x < 420: 
  něco()
```
### Tohle je TAKY SPRÁVNĚ:
```python
x = 42
if x < 420: 
  něco()
elif x > 69:
  něco_jiného()
```
