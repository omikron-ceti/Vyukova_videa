1. **Sudé-liché.**  
Napište program, který zjistí a vypíše, zda je zadané číslo sudé nebo liché.

```python
cislo = int(input("Zadejte cele cislo: "))
if cislo % 2 == 0:
    print(f"Cislo {cislo} je sude.")
else:
    print(f"Cislo {cislo} je liche.")
```

```python
# s vyuzitim prevodu cisla na bool
cislo = int(input("Zadejte cele cislo: "))
if cislo % 2:
    print(f"Cislo {cislo} je liche.")
else:
    print(f"Cislo {cislo} je sude.")
```

2. **Je větší.**  
Napište program, který zjistí a vypíše, které ze dvou zadaných čísel je větší.

```python
cislo1 = float(input("Zadejte cislo: "))
cislo2 = float(input("Zadejte dalsi cislo: "))

if cislo1 - cislo2 > 0: 
    print(f"Cislo {cislo1} je vetsi nez {cislo2}.")
elif cislo1 - cislo2 < 0:
    print(f"Cislo {cislo2} je vetsi nez {cislo1}.")
else: print("Obe zadana cisla jsou stejna.")
```
