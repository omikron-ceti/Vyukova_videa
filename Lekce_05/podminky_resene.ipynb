1. **Sudé-liché.**  
Napište program, který zjistí a vypíše, zda je zadané číslo sudé nebo liché.

```python
cislo = int(input("Zadejte cele cislo: "))
if cislo % 2 == 0:
    print(f"Cislo {cislo} je sude.")
else:
    print(f"Cislo {cislo} je liche.")
```

2. **Je větší.**  
Napište program, který zjistí a vypíše, které ze dvou zadaných čísel je větší.

```python
a = int(input("Zadejte cislo"))
b = int(input("Zadejte dalsi cislo"))
if a > b: 
    print(f"Cislo {a} je vetsi nez cislo {b}")
elif a == b:
    print("Obe zadana cisla jsou stejne velka")
else: 
    print(f"Cislo {b} je vetsi nez cislo {a}")
```

3. **Je největší.**  
Napište program, který zjistí a vypíše, které ze tří zadaných čísel je největší.

```python
a = int(input("Zadejte cislo"))
b = int(input("Zadejte dalsi cislo"))
c = int(input("Zadejte jeste dalsi cislo"))
if a >= b and a >= c: 
    print(f"Cislo {a} je nejvetsi")
else: # ted uz vim ze a neni nejvesi
    if b >= c: 
        print(f"Cislo {b} je nejvetsi")
    else: 
        print(f"Cislo {c} je nejvetsi")
```

4. **Seřaď.**  
Napište program, který vypíše tři zadaná čísla v pořadí od největšího po nejmenší.

```python
nejvetsi = int(input("Zadejte cislo"))
prostredni = int(input("Zadejte dalsi cislo"))
nejmensi = int(input("Zadejte jeste dalsi cislo"))

if prostredni > nejvetsi: 
    new_nejvetsi = prostredni
    prostredni = nejvetsi
    nejvetsi = new_nejvetsi
if nejmensi > prostredni:
    new_prostredni = nejmensi
    nejmensi = prostredni
    prostredni = new_prostredni
if prostredni > nejvetsi: 
    new_nejvetsi = prostredni
    nejvetsi = prostredni
    nejvetsi = new_nejvetsi

print(nejmensi, prostredni, nejvetsi)
# stejna cisla napise vedle sebe

```

```python
# alternativne pomoci listu, funguje pro libovolny pocet cisel
input_string = input("Zadejte nekolik cisel oddelenych mezerou")
string_arr = input_string.split() # timhle rozdelim zadana cisla do jednotlivych prvku
# prvky takhle rozdeleneho vstupu ale budou typu string

unsorted_list = [int(cislo) for cislo in string_arr] # timhle pretypuju kazdy prvek na int
sorted_set = sorted(set(unsorted_list)) 
# set vybere vzdy jen jeden vyskyt zadaneho cisla. sorted seradi cisla v setu

print(sorted_set)
# pokud zadate vice stejnych cisel, budou ve vystupu jen jednou 
```

```python
# bubble sort - spoiler: pouziti cyklu
input_string = input("Zadejte nekolik cisel oddelenych mezerou")
string_arr = input_string.split()
num_arr = [int(cislo) for cislo in string_arr] 

# iterovani listem
for n in range(len(num_arr) - 1):
    # postupne porovnavani prvku vedle sebe
    for i in range(n):
        if num_arr[i] > num_arr[i + 1]: # pokud je soucasny prvek vetsi nez nasledujici, musim je prohodit
            num_arr[i], num_arr[i + 1] = num_arr[i + 1], num_arr[i] # elegantnejsi metoda prohazovani prvku
print(num_arr)
# stejne hodnoty budou vedle sebe. Pokud se chceme zbavit duplicit, je mozne pouzit set jako v predchozim pripade.
```

5. **Trojúhelník.**  
Napište program, který zjistí a vypíše, zda lze sestrojit trojúhelník zadaný pomocí tří délek stran.

```python
a = int(input("Zadejte cislo"))
b = int(input("Zadejte dalsi cislo"))
c = int(input("Zadejte jeste dalsi cislo"))

if a+b > c and a+c > b and b+c > a:
    print("Yep")
else: print("Nope")
```
