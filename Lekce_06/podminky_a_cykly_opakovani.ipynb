1. **Kvadratická rovnice**  
Napište program pro vyřešení kvadratické rovnice $ax^2+bx+c=0$, kde čísla $a$,$b$,$c$ jsou reálné parametry.

```python
from math import sqrt
# uplne nejzakladneji
# vstupní data - zadá uživatel
a = float(input("Zadejte koeficient druhe mocniny"))
b = float(input("Zadejte koeficient prvni mocniny"))
c = float(input("Zadejte konstantu"))

# řešení úlohy - musí fungovat pro libovolné vstupní hodnoty
if a == 0:
    x = -c/b
    print(f"Jedine reseni x = {x}")
else:
    d = b**2 - 4*a*c
    if d>0:
        x1 = (-b + sqrt(d)) / (2*a)
        x2 = (-b - sqrt(d)) / (2*a)
        print(f"Dva koreny: x1 = {x1}, x2 = {x2}")
    elif d<0:
        re_x = -b / (2*a)
        im_x = abs(d)**(1/2) / (2*a)
        print(f"Dva komplexni koreny: x1 = {re_x}+{im_x}i, x2 = {re_x}-{im_x}i")
    else:
        x = -b / (2*a)
        print(f"Dvojity koren x1 = x2 = {x}")
```

 2. **BMI podruhé**  
Napište program, který spočítá BMI uživatele a vypíše, zda má nadváhu, podváhu nebo normální váhu (normální váze odpovídá BMI mezi 18.5 a 25).
Dále program napíše uživateli, jaké je jeho rozmezí optimální hmotnosti.   


```python
vaha = float(input("Zadejte svou vahu v kg"))
vyska = float(input("Zadejte svou vysku v m"))
bmi = vaha / vyska**2
bmi_min = 18.5
bmi_max = 25
idealni_vaha_min = vyska**2 * bmi_min
idealni_vaha_max = vyska**2 * bmi_max
if bmi < bmi_min:
    print(f"Vase BMI je {bmi}. Mate podvahu.")
elif bmi > bmi_max:
    print(f"Vase BMI je {bmi}. Mate nadvahu.")
else:
    print(f"Vase BMI je {bmi}. Mate normalni vahu.")
print(f"Vase optimalni hmotnost je v rozmezi {idealni_vaha_min} kg az {idealni_vaha_max} kg.")
```

3. **Ověření prvočíselnosti**  
Napište program, který ověří, jestli je vstupní celé číslo n prvočíslo nebo ne.   


```python
from math import sqrt

n = int(input("zadej prirozene cislo"))
delitel_nalezen = False
for i in range(2, int(sqrt(n))+1):
    if not n % i: # zbytek po deleni je nula 
        			# a tedy i je delitel n
        print(f"{n} neni prvocislo")
        delitel_nalezen = True
        break;
if not delitel_nalezen:
    print(f"{n} je prvocislo")
```

```python
# kupodivu na to existuje funkce
from sympy import isprime 

n = int(input("Zadejte prirozene cislo"))
if isprime(n):
    print(f"{n} je prvocislo")
else: print(f"{n} neni prvocislo")
```

4. **Ciferný součet**  
Napište program, který sečte jednotlivé cifry zadaného celého kladného celého čísla.   


```python
cislo = int(input("Zadejte kladne cele cislo"))
ciferny_soucet = 0
while cislo >= 1:
    ciferny_soucet += cislo % 10 # prictu posledni cislici
    cislo = cislo//10 # useknu posledni cislici
print(ciferny_soucet)
```

5. **Collatzova posloupnost**  
Napište program, který vypíše Collatzovu posloupnost (do dosažení hodnoty 1) pro dané počáteční přirozené číslo n.Collatzova posloupnost: vezmeme celé kladné číslo n. Pokud je n sudé, vydělíme ho dvěma, pokud je n liché, vynásobíme ho třemi a přičteme jedničku, postup opakujeme.Hypotéza je, že posloupnost pro každé přirozené n nakonec dojde k číslu 1.   


```python
n = int(input("Zadejte kladne cele cislo"))
print(n)
while n != 1:
    if n % 2: # n liche
        n = 3*n + 1
    else:
        n //= 2 # muzu protoze n je sude
    print(n)
```

6. **Prvočíselný rozklad**  
Napište program, který vypíše prvočíselný rozklad přirozeného čísla n.

```python
import sympy # dokumentace k sympy https://docs.sympy.org/latest/modules/ntheory.html#ntheory-functions-reference
from math import sqrt

cislo = int(input("zadejte cislo"))
if sympy.isprime(cislo):
    print(f"cislo {cislo} uz je prvocislo") # timhle si usetrime nejakou praci
prvocisla = list(sympy.primerange(2, cislo)) # tohle vytvori list prvocisel az do zadaneho cisla 

delitele = []
for prvocislo in prvocisla:
    if not cislo % prvocislo:
        delitele.append(prvocislo)
                
print(delitele)
```
