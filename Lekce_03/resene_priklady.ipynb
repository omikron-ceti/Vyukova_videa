1. Napište program, který se uživatele zeptá na jeho hmotnost a pomocí průměrné hustoty lidského těla ($1000~kg/m^3$) spočítá jeho objem. Jaký by měl uživatel průměr, pokud by měl tvar koule?

<img src="objem_koule.jpg" width="10%"/>



```python
from math import pi

hmotnost = float(input("Zadejte svou hmotnost v kg: "))
hustota = 1000 # kg/m^3
objem = hmotnost / hustota # m^3

prumer = 2 * (objem*3 / (4*pi))**(1/3) # m
print(f"Pokud byste měli tvar koule, váš průměr by byl {prumer*100:.1f} cm")
```

2. **Objemy a povrchy geometrických těles**
* Napište program, který spočítá objem a povrch koule o průměru d, který zadá uživatel.
* Napište program, který spočítá objem a povrch čtyřbokého jehlanu se základnou o straně délky a a výškou délky v (viz obrázek).
* Napište program, který spočítá obvod a plochu pravidelného mnohoúhelníku o straně délky a ([vzoreček hír](https://cs.wikipedia.org/wiki/Pravideln%C3%BD_mnoho%C3%BAheln%C3%ADk)).

<img src="jehlan.jpg" width="30%"/>

```python
from math import pi

prumer_koule = float(input("Zadejte průměr koule v cm: "))
objem_koule = 4/3 * pi * (prumer_koule/2)**3
povrch_koule = 4 * pi * (prumer_koule/2)**2
print(f"Objem koule je {objem_koule:.2f} cm^3, povrch koule je {povrch_koule:.2f} cm^2")
```

```python
from math import sqrt

zakladna_a = float(input("Zadejte délku strany základny jehlanu v cm: "))
vyska_v = float(input("Zadejte výšku jehlanu v cm: "))
plocha_podstavy = zakladna_a**2
objem_jehlanu = (plocha_podstavy * vyska_v) / 3

vyska_steny = sqrt(vyska_v**2 + (zakladna_a/2)**2)
jedna_stena = (zakladna_a * vyska_steny) / 2
povrch_jehlanu = plocha_podstavy + 4 * jedna_stena
print(f"Objem jehlanu je {objem_jehlanu:.2f} cm^3, povrch jehlanu je {povrch_jehlanu:.2f} cm^2")
```

```python
from math import pi, tan

pocet_stran = int(input("Zadejte počet stran mnohoúhelníku (minimálně 3): "))
strana_a = float(input("Zadejte délku strany mnohoúhelníku v cm: "))
obvod = pocet_stran * strana_a  
plocha = 1/4 * pocet_stran * strana_a**2 / tan(pi/pocet_stran)
print(f"Obvod mnohoúhelníku je {obvod:.2f} cm, plocha mnohoúhelníku je {plocha:.2f} cm^2")
```

3. **Pravidelné mnohostěny** 

Napište program, který spočítá objem a povrch 
* pravidelného čtyřstěnu, 
* krychle, 
* pravidelného osmistěnu, 
* pravidelného dvanáctistěnu a 
* pravidelného dvacetistěnu

[Vzorečky hír](https://cs.wikipedia.org/wiki/Plat%C3%B3nsk%C3%A9_t%C4%9Bleso)

```python
from math import sqrt

strana_a = float(input("Zadejte délku hrany mnohostěnu v cm: "))

# Pravidelný čtyřstěn (tetraedr) - stěny jsou rovnostranné trojúhelníky
povrch_ctyrsten = sqrt(3) * strana_a**2
objem_ctyrsten = strana_a**3 * sqrt(2) / 12

# Pravidelný šestistěn (krychle) - stěny jsou čtverce
povrch_krychle = 6 * strana_a**2
objem_krychle = strana_a**3

# Pravidelný osmistěn - stěny jsou rovnostranné trojúhelníky
povrch_osmisten = 2 * sqrt(3) * strana_a**2
objem_osmisten = strana_a**3 * sqrt(2) / 3

# Pravidelný dvanáctistěn (dodekaedr) - stěny jsou pravidelné pětiúhelníky
povrch_dvanactisten = 3 * sqrt(25 + 10*sqrt(5)) * strana_a**2
objem_dvanactisten = strana_a**3 * (15 + 7*sqrt(5)) / 4

# Pravidelný dvacetistěn (ikosaedr) - stěny jsou rovnostranné trojúhelníky
povrch_dvacetisten = 5 * sqrt(3) * strana_a**2
objem_dvacetisten = 5 * strana_a**3 * (3 + sqrt(5)) / 12

print(f"Pravidelný čtyřstěn: povrch = {povrch_ctyrsten:.2f} cm^2, objem = {objem_ctyrsten:.2f} cm^3")
print(f"Pravidelný šestistěn: povrch = {povrch_krychle:.2f} cm^2, objem = {objem_krychle:.2f} cm^3")
print(f"Pravidelný osmistěn: povrch = {povrch_osmisten:.2f} cm^2, objem = {objem_osmisten:.2f} cm^3")
print(f"Pravidelný dvanáctistěn: povrch = {povrch_dvanactisten:.2f} cm^2, objem = {objem_dvanactisten:.2f} cm^3")
print(f"Pravidelný dvacetistěn: povrch = {povrch_dvacetisten:.2f} cm^2, objem = {objem_dvacetisten:.2f} cm^3")
```
