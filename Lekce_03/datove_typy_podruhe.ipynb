Už jsme si řekli, k čemu jsou dobré datové typy. Pojďme se nyní blíže podívat na jejich využití

### Funkce `input`
samotná funkce `input` vám vždy vrátí text (typ `str`), bez ohledu, co do ní napíšete. Tedy pokud použijete samotný input k zadání čísla, nebudete s ním moci pracovat jako s číslem. 

Abychom změnili datový typ dat zadaných pomocí funkce `input`, právě třeba na číslo, musíme použít funkci pro konverzi datových typů, například takto:

```python
a = input("Zadej číslo:") # tohle vrátí typ str, tedy ne číslo - to nechceme
b = int(input("Zadej celé číslo:")) # tohle už je číslo
c = float(input("Zadej číslo:")) # teď můžeme zadat i desetinné číslo

print(f"Datový typ proměnné a je {type(a)}")
print(f"Datový typ proměnné b je {type(b)}")
print(f"Datový typ proměnné c je {type(c)}")
```

Samozřejmě pokud se v tuto chvíli do proměnné `b` nebo `c` pokusíme "nacpat" něco jiného (například písmeno, nebo v případě proměnné `b` desetinné číslo), program skončí s chybou. Do proměnné `a` si můžeme ale psát co chceme.

### Datový typ `bool` a logické operace
Datový typ `bool` reprezentuje pravdivostní hodnoty výrokové logiky. Hodnota `True` je *pravda* a hodnota `False` je *nepravda* (pozor, na začátku musí být velké písmeno). 

Na typ `bool` můžeme konvertovat i čísla - cokoli nenulového bude `True`, nula je `False`.

Nad hodnotami typu `bool` můžeme pak provádět výrokovou logiku, např. pomocí logických operátorů `and`, `or`, `not`

```python
a = True
b = False
c = bool(1) # True
d = bool(0) # False
e = bool(-5) # True - záporné číslo je také nenulové
f = bool(1<2) # True
g = bool(1>2) # False

print(f"a = {a}")
print(f"not a = {not a}")      
print(f"b = {b}")
print(f"a and b = {a and b}")
print(f"a or b = {a or b}")
print(f"bool(1) = {c}")
print(f"bool(0) = {d}")
print(f"bool(-5) = {e}")
print(f"bool(1<2) = {f}")
print(f"bool(1>2) = {g}")
```

### Typ `int`
Datový typ `int` slouží k reprezentaci celých čísel. Při zápisu typu `int` čísla nesmí obsahovat nuly na začátku a mezery, je ale možné použít `_` například pro oddělení tisíců. Pokud do `int` vložíme desetinné číslo, převod odstraní desetinnou část (neřídí se pravidly zaokrouhlování, např. 0.9 se stále převede na 0)

```python
a = 42
b = 69_420
c = int("123") # převede řetězec na číslo
d = int(3.99) # převede desetinné číslo na celé (odřízne desetinnou část)
f = int(-5)

# e = 02 # tohle jsem zakomentovala, protože by to vyhodilo chybu
# g = 123 456 # tohle taky vyhodí chybu
# h = int("ahoj") # tohle taky neprojde

print(f"a = {a}, type(a) = {type(a)}")
print(f"b = {b}, type(b) = {type(b)}")
print(f"c = {c}, type(c) = {type(c)}")
print(f"d = {d}, type(d) = {type(d)}")
print(f"f = {f}, type(f) = {type(f)}")
```

### Zápis čísel v jiných soustavách
Kromě nám dobře známé desítkové soustavy můžeme pro zápis čísel použít jakoukoli jinou soustavu. Pro převod čísla do většiny soustav by bylo potřeba si napsat vlastní skript (nebo najít knihovnu, kde to někdo udělal, aby vy už jste nemuseli). 

Python ale umí pracovat s těmi nejčastějšími: dvojkovou, osmičkovou a šestnáctkovou. 

Jako indikace jiné číselné soustavy se používají prefixy:
* `0b` pro binární
* `0o` pro osmičkovou
* `0x` pro šestnáctkovou

Pro vypsání čísla v jiné soustavě se používají funkce:
* `bin()` 
* `oct()`
* `hex()`

```python
desitkove = 42
binarni = 0b101010
osmičkove = 0o52
sestnactkove = 0x2A

print(f"desítkové 42 = {desitkove}")
print(f"binární 0b101010 = {binarni}")  
print(f"osmičkové 0o52 = {osmičkove}")
print(f"šestnáctkové 0x2A = {sestnactkove}")

print(f"bin(desítkové) = {bin(desitkove)}")
print(f"oct(binární) = {oct(binarni)}")
print(f"hex(osmičkové) = {hex(osmičkove)}")
```

### Typ `float`
Jde o datový typ pro reprezentaci čísel s desetinnými místy. Desetinnou část oddělujeme tečkou. Pokud za tečkou nic není, "přeloží" se to na nulovou desetinou část (např. 3. je totéž jako 3.0), ale datový typ `float` se zachová. Stejně jako u `int` můžeme použít `_` jako oddělovač řádů (i za desetinnou tečkou). 

Pro zápis velkých (nebo malých) čísel je možné použít *exponentový tvar*, pomocí písmena `e` nebo `E` (je to jedno). Například číslo 3*10<sup>8</sup> zapíšeme jako `3e8`.

```python
a = 3.       # to je totéž jako 3.0
b = float("3.14") # převede řetězec na číslo
c = 3e8      # to je 3 * 10^8, tedy 300 000 000
d = 12_34.56_78 # podtržítka jsou povolená i v desetinných číslech
e = 141428e-5 # to je 141428 * 10^-5, tedy 1.41428

print(f"3. = {a}")
print(f"float('3.14') = {b}")
print(f"3e8 = {c}")
print(f"12_34.56_78 = {d}")
print(f"141428e-5 = {e}")
```

### Aritmetické operátory
Takže zpátky do první třídy ZŠ... Předpokládám, že sčítání, odčítání, násobení a dělení asi nemusím představovat. Pojďme si ale i tak zrekapitulovat, jaké aritmetické operátory můžeme v Pythonu použít. 
|Operace|Výsledek|Datový typ výsledku|
|-------|--------|-------------------|
|`x + y`  |součet hodnot `x` a `y`| `int`, pokud jsou `x` a `y` oba typu `int`, `float` pokud je alespoň jedno číslo `float`|
|`x - y`  |rozdíl hodnot `x` a `y`|`int`, pokud jsou `x` a `y` oba typu `int`, `float` pokud je alespoň jedno číslo `float`|
|`x * y`  |součin hodnot `x` a `y`|`int`, pokud jsou `x` a `y` oba typu `int`, `float` pokud je alespoň jedno číslo `float`|
|`x / y`  |podíl hodnot `x` a `y`| vždy `float`|
|`x // y`  |celočíselný podíl hodnot `x` a `y`| vždy `int`|
|`x % y`  |zbytek po celočíselném dělení hodnot `x` a `y`|`int` pokud dělím celá čísla, jinak `float`|
|`-x`  |záporná hodnota `x`|stejný jako datový typ `x`|
|`+x`  |kladná hodnota `x`|stejný jako datový typ `x`|
|`x ** y`  |`x` umocněné na `y`|`int` pokud umocňuji celá čísla, jinak `float`|

### Matematické funkce a konstanty
#### Přímo vestavěné v Pythonu
Python sám o sobě má užitečné funkce pro práci s čísly, jako například `min()`,`max()` a `abs()`
* funkce `min()` "sežere" několik čísel (ať už zadané jako samostatné hodnoty, nebo všechny najednou v nějakém datovém kontejneru - o těch si povíme později), a vrátí vám nejnižší z nich.
* funkce `max()` funguje podobně, jenom vrátí nejvyšší hodnotu.
* funkce `abs()` vrátí absolutní hodnotu daného čísla.

Tyto funkce si poradí nejen s datovými typy `int` a `float`, ale dokonce i s `bool`, kde hodnotu `False` interpretuje jako 0 a hodnotu `True` jako 1.

```python
nejmensi = -7
stredni = 42
nejvetsi = 69
print(f"Nejmenší číslo je {min(nejmensi, stredni, nejvetsi)}")
print(f"Největší číslo je {max(nejmensi, stredni, nejvetsi)}")
print(f"Absolutní hodnota čísla {nejmensi} je {abs(nejmensi)}")

# Co se stane, když použijeme min() a max() na hodnoty typu bool?
print(f"Nejmenší hodnota z (True, False) je {min(True, False)}")
print(f"Největší hodnota z (True, False) je {max(True, False)}")
print(f"Absolutní hodnota hodnoty False je {abs(False)}")
print(f"Absolutní hodnota hodnoty True je {abs(True)}")

# a zamícháme tam ještě čísla
print(f"Nejmenší hodnota z (True, False, {nejmensi}, {stredni}, {nejvetsi}) je {min(True, False, nejmensi, stredni, nejvetsi)}")
print(f"Největší hodnota z (True, False, {nejmensi}, {stredni}, {nejvetsi}) je {max(True, False, nejmensi, stredni, nejvetsi)}")
```

#### Další užitečné funkce, které je nutné importovat
Pravidlo 34 Pythonu říká, že pokud něco rxistuje, někdo na to udělal knihovnu. Se samotným Pythonem se dá udělat spousta věcí, ale abychom si ulehčili život, nemusíme si všechno psát a definovat sami. Do Pythonu je možné importovat mnoho knihoven, které nám můžou pomoci s nejrůznějším typem problémů - statistické zpracování dat, matematické funkce, vykreslování grafů, tvorba grafického rozhraní, opravdu cokoli vás napadne. Některé se vám již nainstalovaly při instalaci Pythonu (math), jiné, ty více specialozované, se musí doinstalovat zvlášť (numpy, pandas, matplotlib). 

```python
import math # importujeme knihovnu math, abychom mohli používat její funkce a konstanty
print(f"Hodnota čísla π je přibližně {math.pi}")
print(f"Hodnota čísla e je přibližně {math.e}")
print(f"Sinus čísla π/2 je {math.sin(math.pi/2)}")
print(f"Kosinus čísla π je {math.cos(math.pi)}")
```

```python
# pokud víme, že budeme používat jen některé funkce z knihovny, můžeme je importovat jednotlivě
from math import pi, sin, cos
print(f"Hodnota čísla π je přibližně {pi}") # už nemusíme psát math.pi
print(f"Sinus čísla π/2 je {sin(pi/2)}") 
print(f"Kosinus čísla π je {cos(pi)}") 
# poznámka: při importování jednotlivých funkcí může dojít ke kolizi jmen, 
# pokud byste měli funkci se stejným názvem v nějaké jiné knihovně 
# nebo jste si sami definovali stejnojmennou funkci
```
