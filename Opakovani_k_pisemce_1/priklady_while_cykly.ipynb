---

**Jakou hodnotu bude mít proměnná `c` po skončení následujícího kódu?**

**A)** 0  
**B)** 1  
**C)** 4  
**D)** 5  
**E)** jinou (napište jakou)  
**F)** program se zacyklí a nikdy neskončí

```python
c = 5
while 5 > c > 0:
    if c % 2 == 0:
        break
    else:
        c -= 1 

print(c)
```

**Jakou hodnotu bude mít proměnná `c` po skončení následujícího kódu?**

**A)** 1  
**B)** 4  
**C)** 5  
**D)** 10  
**E)** jinou (napište jakou)  
**F)** program se zacyklí a nikdy neskončí

```python
c = 5
while c <= 10:
    if c % 2 == 0:
        break
    else:
        c -= 1 

print(c)
```
