**Co vypíše následující program?** (Formátování odpovědi na jeden řádek zanedbejte.)

**A)** 1 2 3 4  
**B)** 1 4  
**C)** 1:4  
**D)** 1 2 3  
**E)** 2 3  
**F)** i i i  
**G)** i i i i  
**H)** nic  
**I)** program se ani nespustí nebo skončí s chybou

```python
for i in range(1:4):
    print(i)
```

Takhle se `range` nepoužívá. Funkce `range` jako argument přijímá jedno, dvě, nebo tři čísla, oddělená čárkou: 
* pokud zadáte jedno číslo, počítá `range` od 0
* pokud zadáte dvě čísla, `range` začne na prvním a po jedné jde až ke druhému
* pokud zadáte tři čísla, funguje třetí číslop jako veloikost kroku

Příklady správného použití:

```python
# jedno číslo
for i in range(5):
    print(i)
```

```python
# dvě čísla
for i in range(5, 10):
    print(i)
```

```python
# tři čísla
for i in range(2, 10, 2):
    print(i)
```
