**Vyberte, které z následujících možností obsahují platný kód v jazyce Python, který načte z modulu `math` konstantu `pi` a funkci `cos`:**

**A)** 
```python
import pi, cos from math
```  
**B)** 
```python 
import math from cos, pi
```  
**C)** 
```python
from pi import math, cos
```  
**D)** 
```python
from math import pi, cos
```  
**E)** 
```python
import pi, cos
```

Zkuste si:

````python
#A)
vyraz = 'import pi, cos from math'
try: exec(vyraz)
except: print(f"{vyraz}: Blbina")
else: print(f"{vyraz}: Modul úšpěšně importován")
````

````python
#B)
vyraz = 'import math from cos, pi'
try: exec(vyraz)
except: print(f"{vyraz}: Blbina")
else: print(f"{vyraz}: Modul úšpěšně importován")
````

````python
#C)
vyraz = 'from pi import math, cos'
try: exec(vyraz)
except: print(f"{vyraz}: Blbina")
else: print(f"{vyraz}: Modul úšpěšně importován")
````

````python
#D)
vyraz = 'from math import pi, cos'
try: exec(vyraz)
except: print(f"{vyraz}: Blbina")
else: print(f"{vyraz}: Modul úšpěšně importován")
````

````python
#E)
vyraz = 'import pi, cos'
try: exec(vyraz)
except: print(f"{vyraz}: Blbina")
else: print(f"{vyraz}: Modul úšpěšně importován")
````
