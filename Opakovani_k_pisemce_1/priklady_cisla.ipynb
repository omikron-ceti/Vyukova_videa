**Co vypíše následující program?**

**A)** 1_2_0_1  
**B)** 1201.  
**C)** 1201.0  
**D)** 0.1201  
**E)** 1201  
**F)** něco jiného (napište co)  
**G)** v programu je syntaktická chyba

````python
print(1_2_0_1.)
````

<details>
  <summary>Klikněte pro vysvětlení</summary>

Podtržítko `_` slouží v číselných datových typech (`int`, `float`, ...) jako symbol pro zlepšení čitelnosti dlouhých čísel, typický příklad je využití jako oddělovač tisíců. Python ho ovšem v kódu ignoruje.

</details>

**Vyberte všechny možnosti, kde proměnná `a` obsahuje komplexní číslo s reálnou částí rovnou 1 a imaginární částí rovnou 2:**

**A)** 
```python
a = 1 + 2i
```  
**B)** 
```python
a = 1 + 2j
```  
**C)** 
```python 
a = complex(1, 2)
```  
**D)** 
```python
a = complex(2, 1)
```  
**E)** 
```python
a = complex(-2, 1).conjugate()
```  
**F)** 
```python
a = complex(1, -2).conjugate()
```

Zkuste si:

````python
# A) 
vyraz = '(1 + 2i)'
try: vysledek = eval(vyraz)
except: print(f"A) {vyraz}: Blbina")
else: print(f"A) {vyraz}: {vysledek} {type(vysledek)}")

# B) 
vyraz = '(1 + 2j)'
try: vysledek = eval(vyraz)
except: print(f"B) {vyraz}: Blbina")
else: print(f"B) {vyraz}: {vysledek} {type(vysledek)}")

# C)
vyraz = 'complex(1, 2)'
try: vysledek = eval(vyraz)
except: print(f"C) {vyraz}: Blbina")
else: print(f"C) {vyraz}: {vysledek} {type(vysledek)}")

# D)
vyraz = 'complex(2, 1)'
try: vysledek = eval(vyraz)
except: print(f"D) {vyraz}: Blbina")
else: print(f"D) {vyraz}: {vysledek} {type(vysledek)}")

# E)
vyraz = 'complex(-2, 1).conjugate()'
try: vysledek = eval(vyraz)
except: print(f"E){vyraz}: Blbina")
else: print(f"E) {vyraz}: {vysledek} {type(vysledek)}")

# F)
vyraz = 'complex(1, -2).conjugate()'
try: vysledek = eval(vyraz)
except: print(f"F) {vyraz}: Blbina")
else: print(f"F) {vyraz}: {vysledek} {type(vysledek)}")

````

```python
a = 0 + 1i
```  


```python
a = complex(1)
```  

```python
a = i
```  

```python

```

```python
a = 1i
```

**H)** 

````python
# A) 
vyraz = '(0 + 1i)'
try: vysledek = eval(vyraz)
except: print(f"A) {vyraz}: Blbina")
else: print(f"A) {vyraz}: {vysledek} {type(vysledek)}")

# B) 
vyraz = '(0 + 1j)'
try: vysledek = eval(vyraz)
except: print(f"B) {vyraz}: Blbina")
else: print(f"B) {vyraz}: {vysledek} {type(vysledek)}")

# C)
vyraz = 'complex(0, 1)'
try: vysledek = eval(vyraz)
except: print(f"C) {vyraz}: Blbina")
else: print(f"C) {vyraz}: {vysledek} {type(vysledek)}")

# D)
vyraz = 'complex(1)'
try: vysledek = eval(vyraz)
except: print(f"D) {vyraz}: Blbina")
else: print(f"D) {vyraz}: {vysledek} {type(vysledek)}")

# E)
vyraz = 'i'
try: vysledek = eval(vyraz)
except: print(f"E) {vyraz}: Blbina")
else: print(f"E) {vyraz}: {vysledek} {type(vysledek)}")

# F)
vyraz = 'j'
try: vysledek = eval(vyraz)
except: print(f"F) {vyraz}: Blbina")
else: print(f"F) {vyraz}: {vysledek} {type(vysledek)}")

# G)
vyraz = '1i'
try: vysledek = eval(vyraz)
except: print(f"G) {vyraz}: Blbina")
else: print(f"G) {vyraz}: {vysledek} {type(vysledek)}")

# H)
vyraz = '1j'
try: vysledek = eval(vyraz)
except: print(f"H) {vyraz}: Blbina")
else: print(f"H) {vyraz}: {vysledek} {type(vysledek)}")
````

**Jakou hodnotu vypíše následující program?**

**A)** 0x12  
**B)** 0  
**C)** 8  
**D)** 10  
**E)** 12  
**F)** 18  
**G)** něco jiného(napište co)

````python
print(0o12)
````

<details>
  <summary>Klikněte pro vysvětlení</summary>

prefix `0o` značí zápis v **osmičkové soustavě**. Číslo (12)<sub>8</sub> je 1 * 8<sup>1</sup> + 2 * 8<sup>0</sup> = 8 + 2 = 10 v desítkové soustavě.
</details>
