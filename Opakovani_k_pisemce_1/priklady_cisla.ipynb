**Co vypíše následující program?**

**A)** 1_2_0_1  
**B)** 1201.  
**C)** 1201.0  
**D)** 0.1201  
**E)** 1201  
**F)** něco jiného (napište co)  
**G)** v programu je syntaktická chyba

````python
print(1_2_0_1.)
````

Poznámka: Podtržítko `_` slouží v číselných datových typech (`int`, `float`, ...) jako symbol pro zlepšení čitelnosti dlouhých čísel, typický příklad je využití jako oddělovač tisíců. Python ho ovšem v kódu ignoruje.

**Vyberte všechny možnosti, kde proměnná `a` obsahuje komplexní číslo s reálnou částí rovnou 1 a imaginární částí rovnou 2:**

**A)** 
```python
a = 1 + 2i
```  
**B)** 
```python
a = 1 + 2j
```  
**C)** 
```python 
a = complex(1, 2)
```  
**D)** 
```python
a = complex(2, 1)
```  
**E)** 
```python
a = complex(-2, 1).conjugate()
```  
**F)** 
```python
a = complex(1, -2).conjugate()
```

Zkuste si:

````python
# A) 
try: vysledek = eval('(1 + 2i)')
except: print("A) Blbina")
else: print(f"A) {vysledek} {type(vysledek)}")

# B) 
try: vysledek = eval('(1 + 2j)')
except: print("B) Blbina")
else: print(f"B) {vysledek} {type(vysledek)}")

# C)
try: vysledek = eval('complex(1, 2)')
except: print("C) Blbina")
else: print(f"C) {vysledek} {type(vysledek)}")

# D)
try: vysledek = eval('complex(2, 1)')
except: print("D) Blbina")
else: print(f"D) {vysledek} {type(vysledek)}")

# E)
try: vysledek = eval('complex(-2, 1).conjugate()')
except: print("E) Blbina")
else: print(f"E) {vysledek} {type(vysledek)}")

# F)
try: vysledek = eval('complex(1, -2).conjugate()')
except: print("F) Blbina")
else: print(f"F) {vysledek} {type(vysledek)}")

````

**Jakou hodnotu vypíše následující program?**

**A)** 0x12  
**B)** 0  
**C)** 8  
**D)** 10  
**E)** 12  
**F)** 18  
**G)** něco jiného(napište co)

````python
print(0o12)
````

prefix `0o` značí zápis v osmičkové soustavě. Číslo (12)<sub>8</sub> je 1 * 8<sup>1</sup> + 2 * 8<sup>0</sup> = 8 + 2 = 10 v desítkové soustavě
