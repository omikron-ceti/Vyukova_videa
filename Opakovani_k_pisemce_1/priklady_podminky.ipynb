**Vyberte možnosti, které jsou ekvivalentní následujícímu fragmentu kód pro libovolnou hodnotu x:**

<blockquote>


```python
answer = 0
if x >= 0:
    answer = 1
else: 
    answer = -1
```
</blockquote>

**A)** 
```python
answer = x >= 0 ? + : -1
```  
**B)** 
```python
answer = 1 if x >= 0 else -1
```  
**C)** 
```python
answer = -1 if x >= 0 else 1
```  
**D)** 
```python
answer = 1 while x >= 0 else -1
```  
**E)** 
```python
answer = 1 for x in range (-1, 1)
```  
**F)** žádná z uvedených možností


Woah tohle je komplikovaný, tak pojďme popořadě:

````python
import random # pro volbu náhodného x

# tohle je kód, se kterým porovnáváme
x = random.randint(-10, 10) # volím náhodné x, může být kladné i záporné
answer = 0
if x >= 0:
    answer = 1
else: 
    answer = -1

print(f"x = {x}, answer = {answer}")
````

**Odpověď A:** 

Tohle vypadá divně, že jo? Buď je to nějaká pokročilá Pyhon černá magie, nebo je to blbost. Druhá možnost je správně :)  

````python
# A)
answer = x >= 0 ? + : -1
print(f"x = {x}, answer = {answer}")
````

**Odpověď B:**

podmínku tvaru
```python
if nějaká_podmínka:
    hodnota = výraz1
else:
    hodnota = výraz2
```
lze opravdu nahradit **podmíněným výrazem** 
```python
hodnota = výraz1 if nějaká_podmínka else výraz2
```

````python
# B) 
answer = 1 if x >= 0 else -1
print(f"x = {x}, answer = {answer}")  
````

**Odpověď C:**

Tohle je syntakticky správně, akorát to dělá pravý opak, než chceme

````python
# C) 
answer = -1 if x >= 0 else 1
print(f"x = {x}, answer = {answer}")
````

**Odpověď D:**

Ani ve snu. Příkaz `while` se používá pro cykly.

````python
# D) 
answer = 1 while x >= 0 else -1
print(f"x = {x}, answer = {answer}")
````

**Odpověď E:**

Tohle je ten samý problém jako předtím. s malou změnou by se to ovšem dalo použít k vygenerování listu, který obsahuje dvě jedničky. Ale to bychom předbíhali

````python
# E) 
answer = 1 for x in range (-1, 1)
print(f"x = {x}, answer = {answer}")
````

Pro fajnšmekry: vygenerování listu pomocí for cyklu 

````python
answer = [1 for x in range (-1, 1)]
print(f"x = {x}, answer = {answer}")
````

**Co vypíše následující program?** (Formátování odpovědi na jeden řádek zanedbejte)

**A)** A  
**B)** B  
**C)** A B  
**D)** nic  
**E)** program se ani nespustí nebo skončí s chybou

````python
x = 4
if x < 16 ** 1 / 2:
    print("A")
if x / 3 == x // 3:
    print("B")
````

---
Poznámka: **rozdíl mezi `if` a `elif`:** 

Zkusme se teď zamyslet, co by se stalo, kdyby i druhý výrok byl platný. dejme tomu, že naše podmínky (pro `x = 4`) budou: 
> `x < 16 / 2`  
> `x / 2 == x // 2`

Jediný rozdíl bude ten, že v jednom případě budeme v kódu mít dva `if` po sobě a ve druhém případě použijeme `if`-`elif` řetězec. 

````python
x = 4
if x < 16/2:
    print("A")
if x/2 == x//2:
    print("B")
````

````python
x = 4
if x < 16/2:
    print("A")
elif x/2 == x//2:
    print("B")
````

**Vysvětlení**
  
Příkaz `if` nám "zakládá" nový rozhodovací řetězec, podmínky se tedy vyhodnocují nezávisle na sobě. 

`elif` je naopak součástí jednoho rozhodovacího řetězce, tedy pokud se nějaká předchozí podmínka vyhodnotí jako pravdivá, další `elif` se už nekontrolují.
