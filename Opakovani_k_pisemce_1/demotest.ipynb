### Základy
Co vypíše následující program?  
**A)** Hello, world!  
**B)** "Hello, world!"  
**C)** "Hello", "World!"  
**D)** Hello world!

````python
print("Hello","world!")
````

**Vysvětlení:**  
**A)** Čárka se nevypíše, ta slouží funkci `print` jen jako oddělovač jednotlivých věcí, které vypisuje. Pokud byste chtěli vypsat i čárku, museli byste napsat něco jako:
```python
print("Hello, world!")
``` 
anebo o něco ošklivěji (ne vážně, proč byste to dělali):
```python
print("Hello", ",", "world!") 
```
pozn. tato druhá metoda by vám ale hodila mezeru před čárku (i za čárku samozřejmě, ale tam ji chceme).



**B)** + **C)** Uvozovky se nevypíší, protože ty jen funkci `print` říkají, že se jedná o text. Pokud byste chtěli vypsat uvozovky, musel by kód vypadat např. takto:
```python
print('"Hello, world!"') # varianta B)
```
```python
print('"Hello,"', '"world!"') # varianta C)
```

### Proměnné a výrazy
Jakou hodnotu vypíše následující program?  
**A)** 2  
**B)** 3  
**C)** 4  
**D)** 5

````python
f = min
g = f
f = max
h = max

vysledek = f(2, g(h(1, 5), 3), 4)
print(vysledek)
````

**Vysvětlení**  
V pythonu je možné do proměnných ukládat nejen data (čísla, text, ...), ale i funkce. Tomu se říká **aliasing**.  

Nechť `a` je proměnná a `foo()` je funkce.  
Potom výrazem `a = foo` přiřadíme funkci `foo` tzv. **alias**. S aliasem můžeme pracovat stejně, jako s původní funkcí. 

V našem konkrétním případě:

 ```python 
 f = min    # f nyní dělá totéž, co min
 g = f      # g nyní dělá totéž, co f, tedy g = min
 f = max    # teď už f není min, ale max
 h = max    # h dělá totéž, co max

 vysledek = f(2, g(h(1, 5), 3), 4)
 # toto je ekvivalentní zápisu:
 vysledek = max(2, min(max(1, 5),3),4)
 ```

### Syntaxe, syntaktické chyby
Co vypíše následující program? (Formátování odpovědí na jeden řádek zanedbejte.)  
**A)** 1 2 3 4  
**B)** 1 4  
**C)** 1 2 3  
**D)** i i i  
**F)** i i i i  
**G)** nic  
**H)** program se ani nespustí nebo skončí s chybou

````python
for i in range(1, 4):
    print(i)
    else:
        continue
````

**Vysvětlení:**  
příkaz `else` není možné použít takto samostatně. tento příkaz **musí vždy následovat po `if`**, v rámci nějakého řetězce rozhodnutí: 
``` python
if podmínka:
    Něco udělej
else: 
    Udělej něco jiného
``` 
Případně pro více možností:
``` python
if podmínka:
    Něco udělej
elif jiná podmínka:    # elif můžete mít v rozhodovacím řetězci kolik chcete
    Udělej něco jiného 
else: 
    Udělej ještě něco jiného
```

### Čísla
Jakou hodnotu vypíše následující program?  
**A)** 0x0b  
**B)** 0  
**C)** b  
**D)** 0b  
**E)** 11  
**F)** 12  
**G)** něco jiného (napište co)

````python
print(0x0b)
````

**Vysvětlení:**  
Kromě nám dobře známé desítkové soustavy můžeme pro zápis čísel použít jakoukoli jinou soustavu. Pro převod čísla do většiny soustav by bylo potřeba si napsat vlastní skript (nebo najít knihovnu, kde to někdo udělal, aby vy už jste nemuseli). 

Python ale umí pracovat s těmi nejčastějšími: **dvojkovou, osmičkovou a šestnáctkovou**. 

Jako indikace jiné číselné soustavy se používají prefixy:
* `0b` pro binární
* `0o` pro osmičkovou
* `0x` pro šestnáctkovou 

Tedy zápis `0x0b` nám říká, že se jedná o zápis v **šestnáctkové soustavě**. Do šestnáctkové soustavy převedeme číslo takto:

### Math
Vyberte, které z následujících možností obsahují platný kód v jazyku Python

````python
# A)
include math 
````

````python
# B)
from math include sqrt 
````

````python
# C) 
from math import sin, cos, pi 
````

````python
# D)
import sin, cos from math 
````

````python
# E)
from math import *
````

**Vysvětlení:**  
**A)** `include` není slovo pro import knihovny. Správné slovo by bylo **`import`** (Pokud programujete v C/C++, může se vám to plést)  
**B)** Fujky  
**C)** Tohle je správně  
**D)** Takhle obráceně to napsat nejde  
**E)** Tohle je správně  

**Poznámka k používání funkcí z importovaných knihoven:**
* Existuje ještě možnost `import math`, která udělá podobnou věc, jako `from math import *`, ovšem s jedním rozdílem: 
  * Pokud použijete `import math`, musíte při použití funkcí z této knihovny použít syntax `math.funkce`, např `math.sin()`
  * Pokud použijete `from math import *`, už při volání funkcí nemusíte (a ani nesmíte) referovat na název knihovny, tedy např. místo `math.sin()` použijte jen `sin()`. **Bacha na kolize, pokud toto uděláte s knihovnami, které obsahují stejnojmenné funkce!**(např. `math` i `numpy` obsahují funkci `sin()`) 

* Pokud použijete možnost **C)**, tedy `from math import sin, cos, pi`, máte přístup **jen k těm importovaným funkcím**. Jiné funkce z knihovny `math` nemůžete používat. Ani tady se při používání funkcí nereferuje na název knihovny. 

* Další možností jak importovat knihovu je použití aliasu, například `import numpy as np`. Potom na funkce z knihovny `numpy` referujete pomocí `np`, například `np.array([])`

### While - cykly 
Jakou hodnotu bude mít proměnná `c`po skončení následujícího kódu?  
**A)** 0  
**B)** 1  
**C)** 4  
**D)** 5  
**E)** jinou (napište jakou)  
**F)** program se zacyklí a nikdy neskončí  
**G)** v programu je syntaktická chyba

````python
c = 5
while 5 > c > 0:
    if c % 2 == 0:
        break
    else:
        c -= 1

print(c)
````

**Vysvětlení:**
Hodnota proměnné `c` se nezmění, protože není splněna podmínka pro zahájení cyklu.
