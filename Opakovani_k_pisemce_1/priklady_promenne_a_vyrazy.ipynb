# Proměnné a výrazy
--------------

**Vyberte všechny výrazy, které se vyhodnotí jako nepravdivé:**


**A)**  
```python
bool(42)
```
**B)**  
```python
bool(0)
```
**C)**  
```python
bool(False)
```
**D)**  
```python
(True and False) not True
```
**E)**  
```python
bool(42 % 2)
```
**F)**  
```python
1 < 2 and 2 < 1
```
**G)**  
```python
1 / 2 == 1//2
```

**Zkuste si**

````python
# A)
vyraz = 'bool(42)' 
try: vysledek = eval(vyraz)
except: print(f"A) {vyraz}:  Blbina")
else: print(f"A) {vyraz}: {vysledek}")


# B)
vyraz = 'bool(0)'
try: vysledek = eval(vyraz)
except: print(f"B) {vyraz}: Blbina")
else: print(f"B) {vyraz}: {vysledek}")

# C)
vyraz = 'bool(False)'
try: vysledek = eval(vyraz)
except: print(f"C) {vyraz}: Blbina")
else: print(f"C) {vyraz}: {vysledek}")


# D)
vyraz = '(True and False) not True'
try: vysledek = eval(vyraz)
except: print(f"D) {vyraz}: Blbina")
else: print(f"D) {vyraz}: {vysledek}")

# E)
vyraz = 'bool(42 % 2)'
try: vysledek = eval(vyraz)
except: print(f"E) {vyraz}: Blbina")
else: print(f"E) {vyraz}: {vysledek}")


# F)
vyraz = '1 < 2 and 2 < 1'
try: vysledek = eval(vyraz)
except: print(f"F) {vyraz}: Blbina")
else: print(f"F) {vyraz}: {vysledek}")


# G)
vyraz = '1 / 2 == 1 // 2'
try: vysledek = eval(vyraz)
except: print(f"G) {vyraz}: Blbina")
else: print(f"G) {vyraz}: {vysledek}")

````

**Vyberte všechny výrazy, které se vyhodnotí jako pravdivé:**

**A)**  
```python
bool(42)
```
**B)**  
```python
1 < 42 % 5 < 4
```
**C)**  
```python
bool(True)
```
**D)**  
```python
bool(False)
```
**E)**  
```python
bool(42 % 2)
```
**F)**  
```python
3 ** 2 + 4 ** 2 == 5 * 5
```
**G)**  
```python
1 < 2 or 2 < 1
```

**H)**  
```python
1 // 2 == 0.5
```


````python
# A)
vyraz = 'bool(42)' 
try: vysledek = eval(vyraz)
except: print(f"A) {vyraz}:  Blbina")
else: print(f"A) {vyraz}: {vysledek}")


# B)
vyraz = '1 < 42 % 5 < 4'
try: vysledek = eval(vyraz)
except: print(f"B) {vyraz}: Blbina")
else: print(f"B) {vyraz}: {vysledek}")

# C)
vyraz = 'bool(True)'
try: vysledek = eval(vyraz)
except: print(f"C) {vyraz}: Blbina")
else: print(f"C) {vyraz}: {vysledek}")


# D)
vyraz = 'bool(False)'
try: vysledek = eval(vyraz)
except: print(f"D) {vyraz}: Blbina")
else: print(f"D) {vyraz}: {vysledek}")

# E)
vyraz = 'bool(42 % 2)'
try: vysledek = eval(vyraz)
except: print(f"E) {vyraz}: Blbina")
else: print(f"E) {vyraz}: {vysledek}")


# F)
vyraz = '3 ** 2 + 4 ** 2 == 5 * 5'
try: vysledek = eval(vyraz)
except: print(f"F) {vyraz}: Blbina")
else: print(f"F) {vyraz}: {vysledek}")


# G)
vyraz = '1 < 2 or 2 < 1'
try: vysledek = eval(vyraz)
except: print(f"G) {vyraz}: Blbina")
else: print(f"G) {vyraz}: {vysledek}")

# H)
vyraz = '1 // 2 == 0.5'
try: vysledek = eval(vyraz)
except: print(f"H) {vyraz}: Blbina")
else: print(f"H) {vyraz}: {vysledek}")
````
