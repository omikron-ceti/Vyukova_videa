# Proměnné a výrazy
--------------

**Vyberte všechny výrazy, které se vyhodnotí jako nepravdivé:**

**A)**  
```python
bool(42)
```
**B)**  
```python
bool(0)
```
**C)**  
```python
bool(False)
```
**D)**  
```python
(True and False) not True
```
**E)**  
```python
bool(42 % 2)
```
**F)**  
```python
1 < 2 and 2 < 1
```
**G)**  
```python
1 / 2 == 1//2
```

**Zkuste si**

````python
# A)
try: vysledek = eval('bool(42)')
except: print("A) Blbina")
else: print(f"A) {vysledek}")


# B)
try: vysledek = eval('bool(0)')
except: print("B) Blbina")
else: print(f"B) {vysledek}")

# C)
try: vysledek = eval('bool(False)')
except: print("C) Blbina")
else: print(f"C) {vysledek}")


# D)
try: vysledek = eval('(True and False) not True')
except: print(f"D) Blbina")
else: print(f"D) {vysledek}")

# E)
try: vysledek = eval('bool(42 % 2)')
except: print("E) Blbina")
else: print(f"E) {vysledek}")


# F)
try: vysledek = eval('1 < 2 and 2 < 1')
except: print("F) Blbina")
else: print(f"F) {vysledek}")


# G)
try: vysledek = eval('1 / 2 == 1 // 2')
except: print("G) Blbina")
else: print(f"G) {vysledek}")

````

**Vyberte všechny výrazy, které se vyhodnotí jako pravdivé:**

**A)**  
```python
bool(42)
```
**B)**  
```python
1 < 42 % 5 < 4
```
**C)**  
```python
bool(True)
```
**D)**  
```python
bool(False)
```
**E)**  
```python
bool(42 % 2)
```
**F)**  
```python
3 ** 2 + 4 ** 2 == 5 * 5
```
**G)**  
```python
1 < 2 or 2 < 1
```

**H)**  
```python
1 // 2 == 0.5
```


````python
# A)
try: vysledek = eval('bool(42)')
except: print("A) Blbina")
else: print(f"A) {vysledek}")


# B)
try: vysledek = eval('1 < 42 % 5 < 4')
except: print("B) Blbina")
else: print(f"B) {vysledek}")

# C)
try: vysledek = eval('bool(True)')
except: print("C) Blbina")
else: print(f"C) {vysledek}")


# D)
try: vysledek = eval('bool(False)')
except: print(f"D) Blbina")
else: print(f"D) {vysledek}")

# E)
try: vysledek = eval('bool(42 % 2)')
except: print("E) Blbina")
else: print(f"E) {vysledek}")


# F)
try: vysledek = eval('3 ** 2 + 4 ** 2 == 5 * 5')
except: print("F) Blbina")
else: print(f"F) {vysledek}")


# G)
try: vysledek = eval('1 < 2 or 2 < 1')
except: print("G) Blbina")
else: print(f"G) {vysledek}")

# H)
try: vysledek = eval('1 // 2 == 0.5')
except: print("H) Blbina")
else: print(f"H) {vysledek}")
````
