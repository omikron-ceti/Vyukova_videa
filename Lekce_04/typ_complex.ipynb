## Typ `Complex`
Tento datový typ reprezentuje komplexní čísla. Je možno je zadat jako uspořádanou dvojici hodnot `complex(a, b)`, reprezentující reálnou a imaginární část, nebo pomocí zápisu `a + bj`, kde `a` a `b` jsou nějaká čísla.

**Imaginární jednotka `j` musí být vždy napsaná hned za číslem, jinak si Python bude myslet, že se jedná o proměnnou!**

S komplexními čísly je možné provádět všechny aritmetické operace stejně jako s `int`, **kromě celočíselného dělění `%`**.

**KOMPLEXNÍ ČÍSLA NEJDE POROVNÁVAT!**

```python
komplexni_cislo = complex(1, 2)
dalsi_komplexni_cislo = 3 + 4j
print(f"complex(1, 2) = {komplexni_cislo}")
print(f"3 + 4j = {dalsi_komplexni_cislo}")

j = 42 # menší komplikace - přidám si do kódu proměnnou co se jmenuje j
chaos_a_zmar1 = 3 + j # tady Python vyhodnotí j jako proměnnou a ne jako imaginární jednotku
chaos_a_zmar2 = 3 + 1j # tady to funguje, protože j je hned za číslem
print(f"3 + j = {chaos_a_zmar1}")
print(f"3 + 1j = {chaos_a_zmar2}")
```

Pro přístup k samotné reálné a imaginární části nám slouží tzv. *atributy* `.real` a `.imag`. Použitím těchto atributů dostanete odpovídající části komplexního čísla, které pak budou typu `float`. 

Další užitečný atribut je `conjugate()`, který provede komplexní sdružení čísla.

**Použití atributů nijak nemění obsah původní proměnné, hodnoty získané pomocí atributů je nutno uložit do nové proměnné!**

```python
komplexni_cislo = complex(1, 2)
realna_cast = komplexni_cislo.real
imaginární_cast = komplexni_cislo.imag

print(f"Komplexní číslo je {komplexni_cislo}, jeho reálná část je {realna_cast} a imaginární část je {imaginární_cast}")

komplexni_cislo.conjugate() # tohle sem dávám proto, aby bylo vidět, že použití atributu mi nepřepíše původní hodnotu
sdruzene_cislo = komplexni_cislo.conjugate() # uložím sdružené číslo do nové proměnné
print(f"Komplexní číslo je {komplexni_cislo}, jeho sdružené číslo je {sdruzene_cislo}")
```

 
2. Použijte jazyk Python pro vyřešení kvadratické rovnice $πx^2+2x+1=0$. (Řešení tohoto příkladu nemusí být univerzální program, stačí posloupnost příkazů s výpočty, které byste prováděli "na papíře".)

```python
from math import pi, sqrt
a = pi
b = 2
c = 1
diskriminant = b**2 - 4*a*c
x1 = complex(-b, sqrt(abs(diskriminant))) / (2*a)
x2 = complex(-b, -sqrt(abs(diskriminant))) / (2*a) 
print(f"Rovnice má dva komplecně sdružené kořeny: \nx1 = {x1}\nx2 = {x2}")
```

Pro fajnšmekry: obecné řešení

```python
from math import sqrt
a = float(input("Zadejte kvadratický člen a: "))
b = float(input("Zadejte lineární člen b: "))
c = float(input("Zadejte absolutní člen c: "))

if a == 0:
    print(f"Zadali jste lineární rovnici {b}x + {c} = 0, jejím řešením je x = {-c/b}")

else:
    diskriminant = b**2 - 4*a*c
    if diskriminant > 0:
        x1 = (-b + sqrt(diskriminant)) / (2*a)
        x2 = (-b - sqrt(diskriminant)) / (2*a) 
        print(f"Rovnice má dva reálné kořeny: \nx1 = {x1}\nx2 = {x2}")
    elif diskriminant == 0:
        x1 = -b / (2*a)
        print(f"Rovnice má jeden dvojnásobný kořen: \nx12 = {x1}")
    else:
        x1 = complex(-b, sqrt(abs(diskriminant))) / (2*a)
        x2 = complex(-b, -sqrt(abs(diskriminant))) / (2*a) 
        print(f"Rovnice má dva komplexně sdružené kořeny: \nx1 = {x1}\nx2 = {x2}")
```

Trochu kompaktnější verze


```python
from math import sqrt
from cmath import sqrt as csqrt  # vytvořila jsem alias, abych mohla používat oba druhy odmocnin

a = float(input("Zadejte kvadratický člen a: "))
b = float(input("Zadejte lineární člen b: "))
c = float(input("Zadejte absolutní člen c: "))

if a == 0:
    print(f"Zadali jste lineární rovnici {b}x + {c} = 0, jejím řešením je x = {-c/b}")
else:
    diskriminant = b**2 - 4*a*c
    x1 = (-b + csqrt(diskriminant)) / (2*a)
    x2 = (-b - csqrt(diskriminant)) / (2*a)
    print(f"Řešení rovnice je:\nx1 = {x1}\nx2 = {x2}")

```
