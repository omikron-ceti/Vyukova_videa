````python
def je_sude(n):
    if n % 2 == 0:
        return True
    else: 
        return False

print(je_sude(42))
````

````python
# Protože return končí běh funkce, šlo by to i takto:
def je_sude(n):
    if n % 2 == 0:
        return True
    return False 

print(je_sude(69))
````

````python
# Využití principu 0 <=> False
def je_sude(n):
    if n % 2: # pokud n % 2 == 1, bude výsledek True a číslo není sudé
        return False
    return True 

print(je_sude(34))
````

````python
# Alternativně: využití principu 0 <=> False
def je_sude(n):
    return not n % 2

print(je_sude(67))
````

1. Definujte a zavolejte funkci `je_dělitelné(n, d)`, která vrací `True` nebo `False` podle toho, jestli je zadané přirozené číslo `n` dělitelné číslem `d`.

````python
def je_delitelne(n, d):
    if n % d: # podmínka splněna při nenulovém výsledku operace %
        return False
    else: return True
    
print(je_delitelne(4, 2))
print(je_delitelne(4, 3))
````

````python
# alternativně
def je_delitelne(n, d):
    return not n % d
    
print(je_delitelne(4, 2))
print(je_delitelne(4, 3))
````

 
2. Napište a zavolejte funkci `počet_dělitelů(n)`, která spočítá a vrátí počet dělitelů přirozeného čísla `n`. Využijte funkci naprogramovanou v předchozím příkladu.

````python
def pocet_delitelu(n):
    pocet = 0
    for delitel in range(1, n+1): # n určitě dělí samo sebe, chci jej tak= započítat
        if je_delitelne(n, delitel):
            pocet += 1
    return pocet

print(pocet_delitelu(10))
````

````python
# alternativně: pomocí while cyklu
def pocet_delitelu(n):
    pocet = 0
    delitel = 1
    while delitel <= n:
        if je_delitelne(n, delitel):
            pocet += 1
        delitel += 1
    return(pocet)

print(pocet_delitelu(10))
````

5. Naprogramujte funkci `součet_číslic(n)`, která spočítá součet číslic v desítkovém zápisu přirozeného čísla `n`.

````python
def soucet_cislic(n):
    ciferny_soucet = 0
    while n >= 1:
        ciferny_soucet += n%10  # přečtu poslední cifru a přičtu ji ke stávajícímu součtu
        n //= 10                # zahodím poslední cifru
    return(ciferny_soucet)
    
print(soucet_cislic(1234))
````

6. Naprogramujte funkci `remove(n, digit)`, která z desítkového zápisu zadaného přirozeného čísla `n` odstraní všechny číslice, které se rovnají hodnotě `digit`. Např.:
<blockquote>

  ```python
  remove(123, 2)
  ```
  `output: 13`
</blockquote>
<blockquote>

  ```python
  remove(243132, 3)
  ```
   `output: 2412`
</blockquote>

````python
def remove(n, digit):
    n = str(n) # přetypuji na string, abych mohla použít funkci replace()
    digit = str(digit)
    removed = n.replace(digit, "") # Nahradím všechny výskyty číslice za prázdný string - tím je smažu
    return int(removed) # přetypuji zpět na int

print(remove(123, 2))
print(remove(243132, 3))
````
