# `dict`
---
**Vyberte všechny příkazy, které jsou syntakticky správně a definují proměnnou typu `dict` (slovník):**
```python
a = {1:2}                   e = {}
b = {1:2:3}                 f = {1,2:3,4}
c = {1:2, 3:4}              g = {1:2, 1:3}
d = {1, 2, 3, 4}            h = {1:1, 2:1}
```

````python
# a
vyraz = "{1:2}" 
try: vysledek = eval(vyraz)
except: print(f"A) {vyraz}:  Blbina")
else: print(f"A) {vyraz}: {type(vysledek)}")

# b
vyraz = '{1:2:3}' 
try: vysledek = eval(vyraz)
except: print(f"B) {vyraz}:  Blbina")
else: print(f"B) {vyraz}: {type(vysledek)}")

# c
vyraz = '{1:2, 3:4}' 
try: vysledek = eval(vyraz)
except: print(f"C) {vyraz}:  Blbina")
else: print(f"C) {vyraz}: {type(vysledek)}")

# d
vyraz = '{1, 2, 3, 4}' 
try: vysledek = eval(vyraz)
except: print(f"D) {vyraz}:  Blbina")
else: print(f"D) {vyraz}: {type(vysledek)}")

# e
vyraz = '{}'
try: vysledek = eval(vyraz)
except: print(f"E) {vyraz}:  Blbina")
else: print(f"E) {vyraz}: {type(vysledek)}")

# f
vyraz = '{1,2:3,4}'
try: vysledek = eval(vyraz)
except: print(f"F) {vyraz}:  Blbina")
else: print(f"F) {vyraz}: {type(vysledek)}")

# g
vyraz = '{1:2, 1:3}' 
try: vysledek = eval(vyraz)
except: print(f"G) {vyraz}:  Blbina")
else: print(f"G) {vyraz}: {type(vysledek)}")

# h
vyraz = '{1:1, 2:1}' 
try: vysledek = eval(vyraz)
except: print(f"H) {vyraz}:  Blbina")
else: print(f"H) {vyraz}: {type(vysledek)}")
````

**Co vypíše následující program?** (Parametr `end = " "` ve funkci `print` způsobí výpis na jeden řádek)

**A)** a b c  
**B)** x y z  
**C)** a x b y c z  
**D)** 'a': 'x' 'b': 'y' 'c': 'z'  
**E)** ('a', 'x') ('b', 'y') ('c', 'z')  
**F)** program skončí s chybou, protože slovník nelze použít ve for-cyklu

````python
data = {'a': 'x', 'b': 'y', 'c': 'z'}
for k, v in data.items():
    print(k, v, end=" ")
````

**Co vypíše následující program?** (Parametr `end = " "` ve funkci `print` způsobí výpis na jeden řádek)

**A)** a b c  
**B)** x y z  
**C)** a x b y c z  
**D)** 'a': 'x' 'b': 'y' 'c': 'z'  
**E)** ('a', 'x') ('b', 'y') ('c', 'z')  
**F)** program skončí s chybou, protože slovník nelze použít ve for-cyklu

````python
data = {'a': 'x', 'b': 'y', 'c': 'z'}
for i in data:
    print(i, end=" ")
````
