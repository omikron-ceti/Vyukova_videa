**Doplňte správně chybějící místa v kódu tak, aby program vytvořil mělkou kopii `s` seznamu `t`:**

<blockquote>

```python
t = [1, 2, 3]
s = ...
for ... in ...:
    ...
```

1. možnost:

````python
t = [1, 2, 3]
s = []
for i in range(3):
    s.append(t[i])

print(s) # tohle je tady navíc jen pro výpis
````

Toto je sice správně, ale není to úplně hezké řešení, protože by nefungovalo, kdyby `t` byl jinak dlouhý. Zkusíme to trochu upravit.

2. možnost:

````python
t = [1, 2, 3]
s = []
for i in range(len(t)):
    s.append(t[i])

print(s)
````

O něco lepší, teď by to fungovalo pro jakkoli dlouhý list `t`, ale dá se to udělat ještě líp.

3. možnost:

````python
t = [1, 2, 3]
s = []
for neco in t:
    s.append(neco)

print(s)
t[0] = 42
print(s)
````

<details>
  <summary>Vysvětlení</summary>

Proč s tím děláme takový vylomeniny? Nestačilo by prostě napsat něco jako:
```python

t = [1, 2, 3]
s = t
```

Tohle by na první pohled vypadalo jako správné řešení, ovšem nastal by problém, kdybychom chtěli jeden z listů změnit - změna by se totiž propsala i do kopie. Přiřazením `s = t` totiž nezkopírujeme celý list, ale předáme do proměnné `s` pouze **ukazatel na místo v paměti**, kde je uložen obsah proměnné `t`.

Divný, co? S čísly to takhle přeci nefunguje, pokud by `s` a `t` byly typu `int`, pak `s = t` vytvoří dvě na sobě nezávislé proměnné, a změna jedné tu druhou nijak neovlivní. Rozdíl je v tom, že `int` je takzvaně **immutable** datový typ, kdežto list je **mutable**.

### Mutable vs. immutable datové typy
---
#### Immutable
Mezi **immutable** datové typy patří například `int`, `float`, `str` a `tuple`.

Tyto datové typy jsou **nemodifikovatelné**, což znamená, že v okamžiku, kdy je vytvoříme, nemůžeme jejich hodnotu změnit jinak, než že do nich uložíme nová data. To je divný, vždyť přeci můžu napsat 
```python
t = 3
t = t + 1
```
a změním tím hodnotu uloženou v `t`, tedy proměnná `t` byla změněna. Ovšem k provedení té změny bylo potřeba **přiřadit novou hodnotu pomocí `=`** a tím se efektivně **smazal obsah proměnné a následně se nahradil jiným obsahem. Pokud napíši jen `t+1`, výsledek operace se nikam neuloží!**
Po vykonání řádky 
```python
t = 3
```
bude proměnná `t` ukazovat na místo v paměti, kam program uložil hodnotu 3.
Při vykonání další řádky
```python
t = t+1
```
program nejdříve přečte obsah proměnné `t` (tedy 3), přičte k němu 1, výsledek uloží do paměti a nastaví `t` tak, aby ukazovalo na tento výsledek (který obecně může být uložen na jiné adrese v paměti, než kde byl uložen původní obsah proměnné `t` před přepsáním)

 <img src="img/immutable.jpg" width="40%"/>

Při **kopírování** immutable proměnných pomocí

 ```python
 t = 3
 s = t
 ``` 
sice také budou obě proměnné nejprve odkazovat na stejné místo v paměti:

 <img src="img/mutable.jpg" width="40%"/>

ale pokud se rozhodneme jednu z nich změnit, 
 ```python
s = s + 1
 ``` 
**vložíme do ní novou hodnotu, která je uložená jinde** a tedy přepíšeme ukazatel uložený pouze v této jedné proměnné

 <img src="img/immutable_zmena.jpg" width="40%"/>

 
#### Mutable
Mezi **mutable** datové typy patří `list`, `set` a `dict`. 

Tyto datové typy jsou **modifikovatelné** a jejich obsah tedy můžeme měnit, aniž bychom kvůli tomu přepisovali celou proměnnou. Například pro přidání prvku do listu stačí napsat 
```python
t = [1, 2, 3]
t.append(4)
```
**Vidíte, že jsme pro změnu listu `t` nepotřebovali přiřadit nový obsah pomocí `=`, ale prostě jsme jej "napřímo" změnili.**

 Při **kopírování** mutable proměnných pomocí

 ```python
 t = [1, 2, 3]
 s = t
 ``` 
nekopírujeme obsah, ale **kopírujeme pouze odkaz na uložená data. Změna jedné proměnné se tedy propíše i do její kopie**, protože i ona ukazuje na místo v paměti, které bylo změněno.
 <img src="img/mutable_zmena.jpg" width="40%"/>

Pro vytvoření nezávislé kopie musíme přečíst uložená data a uložit je do nové proměnné. 


### Co to znamená *"mělká kopie"*?
---
Název *"mělká kopie"* implikuje i existenci *"hluboké kopie"* - a je tomu tak. Jaký je v nich rozdíl a k čemu to je? 

**Mělká kopie** znamená, že kopírujeme pouze první "vrstvu" datového kontejneru - například kdybychom měli vícerozměrný list (tedy list obsahující vnořené listy), vytvoří se kopie tím způsobem, že "první úroveň" budou data uložená nezávisle na původní proměnné, ale veškerá "vnořená" data se předávají jenom jako odkazy - a jejich změny se tedy propisují do obou kopií. 

V případě **hluboké kopie** pak kopírujeme úplně všechno, při kopírování je terdy potřeba "vlézt" do kazdé další vnořené struktury a postupně ji také překopírovat. 
</details>

Zkuste si:

````python
# immutable datový typ: int, float, str, tuple
t = 3
s = t
print(f"Immutable datový typ:\n\
        t: {t}\n\
    s = t\n\
        s: {s}")
t = t+1

print(f"\
    t = t + 1\n\
        t: {t}\n\
        s: {s}")


# mutable datový typ: list, set, dict
t = [1, 2, 3]
s = t

print(f"Mutable datový typ:\n\
        t: {t}\n\
    s = t\n\
        s: {s}")

t.append(4)

print(f"\
    t.append(4)\n\
        t: {t}\n\
        s: {s}")
````

Mělká vs. hluboká kopie

````python
def melka_kopie(data):
    kopie = []
    for prvek in data:
        kopie.append(prvek)
    return kopie

def hluboka_kopie(data): # rekurzivně řešená hluboká kopie
    kopie = []
    if isinstance(data, (int, float)): # ověříme, jestli už jsme na nejnižší úrovni struktury
        kopie = data
    else: # pokud je ještě kam pokračovat
        for prvek in data:
            kopie.append(hluboka_kopie(prvek))
    return kopie

nejaky_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
o_kopie = nejaky_list # tady kopírujeme jen odkaz
m_kopie = melka_kopie(nejaky_list)
h_kopie = hluboka_kopie(nejaky_list)

print(f"původní list: {nejaky_list}\n\
mělká kopie: {m_kopie}\n\
hluboká kopie: {h_kopie}\n")
````

````python
if o_kopie is nejaky_list:
    print("Původní list a list vytvořený pomocí '=' jsou uloženy na stejném místě v paměti.")
else: print("Původní list a list vytvořený pomocí '=' jsou uloženy na jiném místě v paměti.")

if m_kopie is nejaky_list:
    print("Původní list a mělká kopie jsou uloženy na stejném místě v paměti.")
else: print("Původní list mělká kopie jsou uloženy na jiném místě v paměti.")

if m_kopie[0] is nejaky_list[0]:
    print("Vnořené listy v originále a v mělké kopii jsou uloženy na stejném místě v paměti.")
else: print("Vnořené listy v originále a v mělké kopii jsou uloženy na jiném místě v paměti.")

if h_kopie is nejaky_list:
    print("Původní list a hluboká kopie jsou uloženy na stejném místě v paměti.")
else: print("Původní list hluboká kopie jsou uloženy na jiném místě v paměti.")

if h_kopie[0] is nejaky_list[0]:
    print("Vnořené listyv originále a v hluboké kopii jsou uloženy na stejném místě v paměti.")
else: print("Vnořené listy v originále a v hluboké kopii jsou uloženy na jiném místě v paměti.")
````

````python
# Změním něco v 1. úrovni
h_kopie.pop(2) # odstraním poslední prvek
m_kopie.pop(0)
print(f"hluboká kopie po odstranění posledního prvku: {h_kopie}\n\
mělká kopie po odstranění prvního prvku: {m_kopie}\n\
původní list: {nejaky_list}\n")

# Změním něco ve 2. úrovni
h_kopie[1].pop(2) # odstraním poslední prvek
print(f"hluboká kopie po změně ve 2. úrovni: {h_kopie}\n\
původní list: {nejaky_list}\n")

m_kopie[0].pop(0)
print(f"mělká kopie po změně ve 2. úrovni: {m_kopie}\n\
původní list: {nejaky_list}\n\
proměnná obsahující kopii odkazu: {o_kopie}")
````

### `tuple`
**Jakou hodnotu bude mít proměnná `b` po spuštění následujícího kódu?**  
**A)** -2  
**B)** -1  
**C)** 0  
**D)** 2  
**E)** 3  
**F)** jinou(uveďte jakou)
**G)** proměnná `b` nebude definována
**G)** program se nespustí, nebo předčasně skončí kvůli chybě


````python
def f(x):
    y = x + 1
    return (y, x - y)

a, b = f(2)

print(b)
````

<details>
  <summary>Vysvětlení</summary>

Tuple je možné "rozbalit" do jednotlivých proměnných právě pomocí syntaxu, který je použitý v tomto příkladu. Samozřejmě těch proměnných musí být právě tolik, kolik je prvků v tuple. 

</details>

````python
# poznámka: rozbalování tuplů
x = (69, 420, 67)
a, b, c = x
print(a, b, c)
````
