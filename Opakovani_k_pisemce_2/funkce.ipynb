**Co vypíše následující funkce?**

**A)** 1 2  
**B)** {1, 2}  
**C)** (1, 2)  
**D)** [1, 2]  
**E)** ((1, 2),)  
**F)** ([1, 2])  
**G)** něco jiného (napište co)

```python
def fun(*args):
    print(*args)

fun(1, 2)
```

Od předchozího příkladu se liší hvězdičkou. Hvězdička znamená, že funkce může přijmout libovolné množství argumentů. Python si je sám uzavře do `tuple`, ale pokud je takto vypisujeme opět pomocí `*`, vypíší se jen jednotlivé hodnoty. Samotné `args` **není klíčové slovo**, argument funkce se může jmenovat jakkoli.

Příklad pro upřesnění:

```python
def fun(*nejake_srandovni_jmeno):
    print("výpis s *:", *nejake_srandovni_jmeno)    # pro výpis s * nelze použít formátovaný řetězec :(
    print(f"výpis bez *: {nejake_srandovni_jmeno}")

fun(1, 2)
```

**Jakou hodnotu vypíše následující program?**

**A)** 3!  
**B)** 6  
**C)** program se zacyklí a nic nevypíše  
**D)** program skončí s chybou kvůli "nekonečné" rekurzi

```python
def f(n):
    return n * f(n-1)

print(f(3))
```

Není nic, co by rekurzi zastavilo, **program skončí s chybou**

**Jakou hodnotu bude mít proměnná `b` po spuštění následujícího kódu?**

**A)** -2  
**B)** -1  
**C)** 0  
**D)** 1  
**E)** 2  
**F)** jinou (uveďte jakou)  
**G)** proměnná `b` nebude definována  
**H)**  program se nespustí nebo předčasně skončí kvůli chybě

```python
def f(x, y=None):
    return min(x, y), max(x, y)

a, b = f(2, -2)

print(b)
```

Do proměnnné `b` bude uložen výsledek výrazu `max(2, -2)`, což je 2.
