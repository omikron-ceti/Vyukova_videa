**Jakou hodnotu bude mít proměnná `y` po spuštění následujícího kódu?**

**A)** 0  
**B)** 1  
**C)** None  
**D)** null  
**E)** proměnná `y` nebude definovaná  
**F)** program skončí kvůli chybě

```python
y = 0

def f(x):
    y = x + 1
    print(y)

f(y)

print(f"Hodnota proměnné y po skončení programu: {y}") 
```

V rámci funkce `f()` si sice definujeme proměnnou s názvem `y`, ale tato proměnná, ačkoli se jmenuje stejně, je nezávislá na proměnné `y`, která byla vytvořena mimo funkci, a po skončení funkce zanikne. 
Proměnnou `y` tedy nijak neměníme.

**Jakou hodnotu bude mít proměnná `y` po spuštění následujícího kódu?**

**A)** 0  
**B)** 1  
**C)** 2 
**D)** 3  
**E)** jinou (uveďte jakou)  
**F)** program skončí kvůli chybě

```python
y = 0

def f(x):
    x = y + 1
    return(x)

y = f(2)

print(f"Hodnota proměnné y po skončení programu: {y}") 
```

Okey, trochu jsem to předtím zjednodušila. Pokud si ve funkci **definujeme** proměnnou, která se jmenuje stejně, jako proměnná použitá v kódu, vytvoříme tím novou, stejnojmennou, ale nezávislou proměnnou. 

Pokud ale ve funkci **voláme, ale nedefinujeme** proměnnou, kterou jsme si definovali někde mimo (před definicí funkce), **pak funkce tuto hodnotu využije**. Proto při definici funkce nedostaneme chybu, že proměnnou `y` funkce nezná. 

**Toto ale není vůbec hezké, obecně bychom ve funkcích nikdy neměli používat proměnné, které jim nepředáme jako argumenty, anebo si je ve funkci nezadefinujeme**. Použití globálních proměnných může vést k nepředvídatelnému chování funkce (např. pokud si danou proměnnou někde "po cestě" změníme).

**Co vypíše následující funkce?**

**A)** 1 2  
**B)** {1, 2}  
**C)** (1, 2)  
**D)** [1, 2]  
**E)** ((1, 2),)  
**F)** ([1, 2])  
**G)** něco jiného (napište co)

```python
def fun(*args):
    print(*args)

fun(1, 2)
```

Od předchozího příkladu se liší hvězdičkou. Hvězdička znamená, že funkce může přijmout libovolné množství argumentů. Python si je sám uzavře do `tuple`, ale pokud je takto vypisujeme opět pomocí `*`, vypíší se jen jednotlivé hodnoty - **hvězdička nám totiž umožňuje přistoupit k vnitřku datového kontejneru**. 

Samotné `args` **není klíčové slovo**, argument funkce se může jmenovat jakkoli.

Příklad pro upřesnění:

```python
def fun(*nejake_srandovni_jmeno):
    print("výpis s *:", *nejake_srandovni_jmeno)    # pro výpis s * nelze použít formátovaný řetězec :(
    print(f"výpis bez *: {nejake_srandovni_jmeno}")

fun(1, 2)
```

**Jakou hodnotu vypíše následující program?**

**A)** -1  
**B)** 0  
**C)** 1  
**D)** 2  
**E)** 3  
**F)** jinou(napište jakou)


```python
def f(n):
    if n == 0:
        return n
    return 1-f(n-1)

print(f(3))
```

**Jakou hodnotu bude mít proměnná `b` po spuštění následujícího kódu?**

**A)** -2  
**B)** -1  
**C)** 0  
**D)** 1  
**E)** 2  
**F)** jinou (uveďte jakou)  
**G)** proměnná `b` nebude definována  
**H)**  program se nespustí nebo předčasně skončí kvůli chybě

```python
def f(x, y=None):
    return min(x, y), max(x, y)

a, b = f(2, -2)

print(b)
```

Do proměnnné `b` bude uložen výsledek výrazu `max(2, -2)`, což je 2.
