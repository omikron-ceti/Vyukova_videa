**Vyberte všechny výrazy, které z dvojice `(1, 2)` vytvoří trojici `(1, 3, 2)`**
<blockquote>

```python
t = (1, 2)
```
</blockquote>

**A)**
```python
(t[0], 3, t[1])
```

**B)**
```python
t.insert(1, 3)
```

**C)**
```python
(t[0] + 3 + t[1])
```

**D)**
```python
(t[0], 3, t[-1])
```
**E)**
```python
(t[:1] + tuple(3) + t[1:])
```

**F)** žádná z uvedených možností

````python
t = (1, 2)

# a
vyraz = "(t[0], 3, t[1])" 
try: vysledek = eval(vyraz)
except: print(f"A) {vyraz}:  Blbina")
else: print(f"A) {vyraz}: {vysledek}, {type(vysledek)}")

# b
vyraz = 't.insert(1, 3)' 
try: vysledek = eval(vyraz)
except: print(f"B) {vyraz}:  Blbina")
else: print(f"B) {vyraz}: {vysledek}, {type(vysledek)}")

# c
vyraz = '(t[0] + 3 + t[1])' 
try: vysledek = eval(vyraz)
except: print(f"C) {vyraz}:  Blbina")
else: print(f"C) {vyraz}: {vysledek}, {type(vysledek)}")

# d
vyraz = '(t[0], 3, t[-1])' 
try: vysledek = eval(vyraz)
except: print(f"D) {vyraz}:  Blbina")
else: print(f"D) {vyraz}: {vysledek}, {type(vysledek)}")

# e
vyraz = '(t[:1] + tuple(3) + t[1:])'
try: vysledek = eval(vyraz)
except: print(f"E) {vyraz}:  Blbina")
else: print(f"E) {vyraz}: {vysledek}, {type(vysledek)}")
````
