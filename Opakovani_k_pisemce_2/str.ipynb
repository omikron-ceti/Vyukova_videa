````python
string = "Toto je velmi záludná otázka"

# a
vyraz = "len(string.split())" 
try: vysledek = eval(vyraz)
except: print(f"A) {vyraz}:  Blbina")
else: print(f"A) {vyraz}: {vysledek}")

# b
vyraz = 'len(string[:])' 
try: vysledek = eval(vyraz)
except: print(f"B) {vyraz}:  Blbina")
else: print(f"B) {vyraz}: {vysledek}")

# c
vyraz = 'len(string[0:-1])' 
try: vysledek = eval(vyraz)
except: print(f"C) {vyraz}:  Blbina")
else: print(f"C) {vyraz}: {vysledek}")

# d
vyraz = 'len(string.count(""))' 
try: vysledek = eval(vyraz)
except: print(f"D) {vyraz}:  Blbina")
else: print(f"D) {vyraz}: {vysledek}")

# e
vyraz = 'len(string.split(" "))'
try: vysledek = eval(vyraz)
except: print(f"E) {vyraz}:  Blbina")
else: print(f"E) {vyraz}: {vysledek}")

# f
vyraz = 'len(string)'
try: vysledek = eval(vyraz)
except: print(f"F) {vyraz}:  Blbina")
else: print(f"F) {vyraz}: {vysledek}")

# g
vyraz = 'len(list(string))' 
try: vysledek = eval(vyraz)
except: print(f"G) {vyraz}:  Blbina")
else: print(f"G) {vyraz}: {vysledek}")

# h
vyraz = 'len(string.size)' 
try: vysledek = eval(vyraz)
except: print(f"H) {vyraz}:  Blbina")
else: print(f"H) {vyraz}: {vysledek}")
````

**Který z následujících výrazů vytvoří z proměnné `s = ['10', '20', '30']` řetězec `'10, 20, 30'`?**

**A)**
```python
s.join(', ')
```
**B)**
```python
s.concat(', ')
```
**C)**
```python
', '.join(s)
```
**D)**
```python
join(s, ', ')
```
**E)**
```python
concat(s)
```
**F)** žádný z uvedených výrazů

````python
s = ['10', '20', '30']
# a
vyraz = "s.join(', ')" 
try: vysledek = eval(vyraz)
except: print(f"A) {vyraz}:  Blbina")
else: print(f"A) {vyraz}: {vysledek}")

s = ['10', '20', '30']
# b
vyraz = "s.concat(', ')"
try: vysledek = eval(vyraz)
except: print(f"B) {vyraz}:  Blbina")
else: print(f"B) {vyraz}: {vysledek}")

s = ['10', '20', '30']
# c
vyraz = "', '.join(s)" 
try: vysledek = eval(vyraz)
except: print(f"C) {vyraz}:  Blbina")
else: print(f"C) {vyraz}: {vysledek}")

s = ['10', '20', '30']
# d
vyraz = "join(s, ', ')" 
try: vysledek = eval(vyraz)
except: print(f"D) {vyraz}:  Blbina")
else: print(f"D) {vyraz}: {vysledek}")

s = ['10', '20', '30']
# e
vyraz = "concat(s)"
try: vysledek = eval(vyraz)
except: print(f"E) {vyraz}:  Blbina")
else: print(f"E) {vyraz}: {vysledek}")
````
