**Spojte správně následující proměnné (v prostředním sloupci) s jejich daotvými typy v Pythonu. Škrtněte možnosti v prostředním sloupci, které nejsou syntakticky správně.** 
```python
bool            a = 1.2         decimal
boolean         b = 3.45        double
compl           c = False       float
complex         d = '42'        hex
digit           e = -0          int
digital         f = 0...0       integer
num             g = 0x45        str
number          h = "-78"       string

```


````python
# a
vyraz = '1.2' 
try: vysledek = eval(vyraz)
except: print(f"A) {vyraz}:  Blbina")
else: print(f"A) {vyraz}: {type(vysledek)}")

# b
vyraz = '3.45' 
try: vysledek = eval(vyraz)
except: print(f"B) {vyraz}:  Blbina")
else: print(f"B) {vyraz}: {type(vysledek)}")

# c
vyraz = 'False' 
try: vysledek = eval(vyraz)
except: print(f"C) {vyraz}:  Blbina")
else: print(f"C) {vyraz}: {type(vysledek)}")

# d
vyraz = "'42'" 
try: vysledek = eval(vyraz)
except: print(f"D) {vyraz}:  Blbina")
else: print(f"D) {vyraz}: {type(vysledek)}")

# e
vyraz = '-0'
try: vysledek = eval(vyraz)
except: print(f"E) {vyraz}:  Blbina")
else: print(f"E) {vyraz}: {type(vysledek)}")

# f
vyraz = '0...0'
try: vysledek = eval(vyraz)
except: print(f"F) {vyraz}:  Blbina")
else: print(f"F) {vyraz}: {type(vysledek)}")

# g
vyraz = '0x45' 
try: vysledek = eval(vyraz)
except: print(f"G) {vyraz}:  Blbina")
else: print(f"G) {vyraz}: {type(vysledek)}")

# h
vyraz = '"-78"' 
try: vysledek = eval(vyraz)
except: print(f"H) {vyraz}:  Blbina")
else: print(f"H) {vyraz}: {type(vysledek)}")
````

**A)** 0

Vysvětlení: zkusíme si vyhodnotit jednotlivé čáasti výroku:

````python
print(f"len({{1: 2}}) = {len({1: 2})}\n\
len({{1, 2}}) = {len({1, 2})}")
````

Vidíme že výsledné délky se nerovnají, výrok je tedy nepravdivý, tedy má hodnotu `False`.

**Spojte správně následující proměnné (v prostředním sloupci) s jejich datovými typy v Pythonu. Škrtněte možnosti v prostředním sloupci, které nejsou syntakticky správně**
```python
arr             a = (1,2)           listing
array           b = '1,2'           tuple
dict            c = {1:2}           pair
dictionary      d = <1,2>           set
str             e = {1, 2}          int
string          f = [1, 2]          list
```

````python
# a
vyraz = '(1,2)' 
try: vysledek = eval(vyraz)
except: print(f"A) {vyraz}:  Blbina")
else: print(f"A) {vyraz}: {type(vysledek)}")

# b
vyraz = "'1,2'" 
try: vysledek = eval(vyraz)
except: print(f"B) {vyraz}:  Blbina")
else: print(f"B) {vyraz}: {type(vysledek)}")

# c
vyraz = '{1:2}' 
try: vysledek = eval(vyraz)
except: print(f"C) {vyraz}:  Blbina")
else: print(f"C) {vyraz}: {type(vysledek)}")

# d
vyraz = '<1,2>' 
try: vysledek = eval(vyraz)
except: print(f"D) {vyraz}:  Blbina")
else: print(f"D) {vyraz}: {type(vysledek)}")

# e
vyraz = '{1, 2}'
try: vysledek = eval(vyraz)
except: print(f"E) {vyraz}:  Blbina")
else: print(f"E) {vyraz}: {type(vysledek)}")

# f
vyraz = '[1, 2]'
try: vysledek = eval(vyraz)
except: print(f"F) {vyraz}:  Blbina")
else: print(f"F) {vyraz}: {type(vysledek)}")
````
